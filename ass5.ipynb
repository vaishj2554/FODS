```python
# First we will convert the data type of message to object
df['Message']=df['Message'].astype('object')

#now we will fill the na values as no message 
df['Message']=df['Message'].fillna("no message")
```

```python
df.drop_duplicates(inplace=True)# drop the duplicate values
```

```python
df['Start_Date']=pd.to_datetime(df['Start_Date'])#convert the date in general format and if null value is there it will place 'NaT'
```

```python
# Data Transformation
```

```python
# Min-Max Normalization
```

```python
df['Normalized Value'] = (df['Data Value'] - df['Data Value'].min()) / (df['Data Value'].max() - df['Data Value'].min())
```

```python
#Now we will seperate the date into year and month column respectively
```

```python
df['Year']=df['Start_Date'].dt.year
```

```python
df['Month']=df['Start_Date'].dt.year
```

```python
#now we will convert the categorical data into numeric data using One hot encoding
```
