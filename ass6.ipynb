```python
# Error Handling
```

```python
# we will check if the RestingBP is zero for any entry as it is practically impossible to have so
anomalous_bp = df[df['RestingBP'] == 0]
print("Entries with RestingBP = 0:\n", anomalous_bp)
```

```python
# replace anomalous values with median
```

```python
df['RestingBP'] = df['RestingBP'].replace(0, df['RestingBP'].median())
```

```python
# we will check if the Cholesterol is zero for any entry as it is practically impossible to have so
```

```python
anomalous_chol=df[df['Cholesterol']==0]
print("Entries with Cholesterol=0:\n",anomalous_chol)
```

```python
# replace anomalous values with 0
```

```python
df['Cholesterol'] = df['Cholesterol'].replace(0, df['Cholesterol'].median())
```

```python
# now check if oldpeak value is negative as it is unrealistic
```

```python
anomalous_oldpeak=df[df['Oldpeak']<0]
print("Entries with Oldpeak<0 :\n",anomalous_oldpeak)
```

```python
#now set the negative values to zero
```

```python
df['Oldpeak'] = df['Oldpeak'].apply(lambda x: max(x, 0))
```

```python
# now we will encode the categorical data into numeric data using one hot encoding
```

```python
heart_data_encoded = pd.get_dummies(df, columns=['Sex', 'ChestPainType', 'RestingECG', 'ExerciseAngina', 'ST_Slope'])
```
